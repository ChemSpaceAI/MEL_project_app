```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import SDWriter
from rdkit.Chem import PandasTools
```

```python
def write_sdf_from_dataframe(df, filename, smiles_column="SMILES", sample_size=None):
    """
    Write molecules from a DataFrame to an SDF file, including other columns as properties.
    """
    if sample_size is not None:
        df_subset = df.sample(sample_size)
    else:
        df_subset = df

    with SDWriter(filename) as writer:
        for _, row in df_subset.iterrows():
            mol = Chem.MolFromSmiles(row[smiles_column])
            if mol:
                for col in df.columns:
                    if col != smiles_column:
                        mol.SetProp(col, str(row[col]))
                writer.write(mol)

    print(f"SDF file '{filename}' generated successfully with {len(df_subset)} molecules!")
```

```python
united_all = pd.read_csv('Generated_MEL_on_FINAL_data/processed_enumerated_MEL/united/united_all.csv')

comp_2 =  united_all[united_all['mel_type'] == '2_component']
comp_3 = united_all[united_all['mel_type'] == '3_component']
brdg = united_all[united_all['mel_type'] == 'bridge']
```

```python
write_sdf_from_dataframe(comp_2, "Test_Input_1/2_comp.sdf", sample_size=5000)
```

```python
write_sdf_from_dataframe(comp_3, "Test_Input_1/3_comp_iter_1.sdf", sample_size=1000)
write_sdf_from_dataframe(brdg, "Test_Input_1/bridge_iter_1.sdf", sample_size=1000)
```

```python
comp_3_iter_1 = pd.read_csv('Test_Output_1/3_comp_iter_1/enumeration_results.csv')
write_sdf_from_dataframe(comp_3_iter_1, "Test_Input_1/3_comp_iter_2.sdf", sample_size=50000, smiles_column='Smiles')
```
